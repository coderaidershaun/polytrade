# Fills of one run

Markouts, effective spread and slices of the run that `POSTTRADE_RUN` names, from `posttrade.metrics.fills`. A markout is in basis points; negative is against us. Run `posttrade check` first: a run with findings has no honest measure.

```python
import os
from pathlib import Path

from posttrade import contract, runs
from posttrade.metrics import fills
```

```python
run = runs.open_run(Path(os.environ["POSTTRADE_RUN"]), contract.load_repo_contract())
measures = fills.measure(run, fills.FillParams(seed=0))
```

```python
measures.metrics
```

```python
measures.slices
```
